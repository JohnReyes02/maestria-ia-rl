La función `expected_reward` recibe como parámetro el brazo a escoger. Esta función se encarga de ejecutarel brazo dado y retorna la recompensa recibida de la ejecución, de acuerdo a la fórmula:

$$Q_t(b) = \dfrac{\sum\limits_{i=1}^{t-1} r_i1_{a_t=b}}{\sum\limits_{i=1}^{t-1} 1_{a_t=b}} $$

En la fórmula, la recompensa del brazo a escoger $b$ en el momento de la ejecución $t$ se define por la suma de las recompensas que se han recibido para el brazo hasta el momento, sobre la sumatoria de las veces que se ha escogido el brazo.
